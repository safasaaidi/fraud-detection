graphique 1: distriibution des transactions dans le temps:
Interprétation
>Le motif se répète toutes les 24 heures. Les deux creux sont à la 5e et à la 28e heure (écart d’environ 24 h), et la montée du matin se produit vers la 9e et la 33e heure. C’est un cycle quotidien.

>Hypothèse (à vérifier) : les creux sont la nuit. Il y a 5 à 8 fois moins de transactions dans les creux que sur les plateaux, ce qui ressemble à un rythme jour/nuit. Mais on ne connaît pas l’heure réelle de la première transaction. Time = 0 n’est peut-être pas minuit.

>La première barre est plus haute que ses voisines (environ 3 900 contre 2 200). Le dataset commence donc probablement pendant une période encore active, qui décroît vers la nuit.

>Le deuxième jour est un peu plus plat : son maximum (environ 8 700) est inférieur à celui du premier jour (environ 9 900).

>Le volume ne tombe jamais à zéro, même la nuit : il y a toujours des transactions (paiements en ligne, autres fuseaux horaires).

# classement par corrélation avec Class

*Constats*

* Les variables V sont presque non corrélées entre elles (heatmap), ce qui est une conséquence de l’ACP : chacune apporte une information différente.
* Neuf variables ressortent avec Class, les plus fortes étant V17 (environ -0,31), V14 (environ -0,29) et V12 (environ -0,25). Amount et Time sont proches de 0.
* Pour V17, la médiane des fraudes est de -5,16 contre -0,07 pour les normales. Les boîtes ne se chevauchent pas (fraudes : -11,59 à -1,13, normales : -0,48 à +0,40).
* Les fraudes sont beaucoup plus dispersées : écart interquartile de 10,46 contre 0,88 (environ 12 fois plus). Leur moyenne (-6,46) est plus basse que leur médiane (-5,16), ce qui indique une queue vers les valeurs très basses (minimum -25,16).
* Sur l’histogramme, 78,4 % des fraudes étaient hors de la plage affichée (contre 0,87 % des normales). La plage choisie sur l’ensemble des données cachait donc l’essentiel des fraudes.

*Limites*

* La corrélation de Pearson ne mesure que les liens linéaires. Un score de -0,31 sous-estime V17, qui sépare surtout par ses valeurs extrêmes.
* Avec 0,17 % de fraudes, aucun score ne peut être élevé : seul le rang compte.
* Les normales ont aussi des valeurs extrêmes (de -17,10 à +9,25) : V17 seule ne suffit pas à détecter une fraude.
* Le classement est calculé sur toutes les données. Il ne servira pas à choisir les variables du modèle.
* Le sens des V est inconnu (ACP) : pas d’explication causale.

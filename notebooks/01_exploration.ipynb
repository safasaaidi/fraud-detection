# 6. visialisation d'une seul axe (univarié)


```python
df["Time"].hist(bins=10)
plt.xlabel("Time")
plt.ylabel("Frequency")
plt.title("Distribution of transaction times")
plt.show()
```

```python
df["Time_h"] = df["Time"] / 3600
plt.figure(figsize=(10,5))
sns.histplot(data=df, x="Time_h",binwidth=1,stat="count",element="bars",fill=True,kde=True
             ,color="green",edgecolor="red",alpha=0.3)
plt.title("Distribution of transaction times")
plt.xlabel("Time (hours)")
plt.ylabel("number of transactions ")
plt.xticks(range(0,49,5))
plt.xticks(rotation=45,color="red")
plt.grid(True,alpha=0.3)
plt.tight_layout()
plt.show()
```

graphique 1: distriibution des transactions dans le temps:
Interprétation
>Le motif se répète toutes les 24 heures. Les deux creux sont à la 5e et à la 28e heure (écart d’environ 24 h), et la montée du matin se produit vers la 9e et la 33e heure. C’est un cycle quotidien.

>Hypothèse (à vérifier) : les creux sont la nuit. Il y a 5 à 8 fois moins de transactions dans les creux que sur les plateaux, ce qui ressemble à un rythme jour/nuit. Mais on ne connaît pas l’heure réelle de la première transaction. Time = 0 n’est peut-être pas minuit.

>La première barre est plus haute que ses voisines (environ 3 900 contre 2 200). Le dataset commence donc probablement pendant une période encore active, qui décroît vers la nuit.

>Le deuxième jour est un peu plus plat : son maximum (environ 8 700) est inférieur à celui du premier jour (environ 9 900).

>Le volume ne tombe jamais à zéro, même la nuit : il y a toujours des transactions (paiements en ligne, autres fuseaux horaires).

> Ce que je constate (jeu de données sans doublons)

- Le montant médian d’une fraude est plus bas que celui d’une transaction normale : 9,82 € contre 22,00 €.
- Mais la moyenne va dans l’autre sens : 123,87 € pour les fraudes contre 88,41 € pour les normales. Cet écart entre moyenne et médiane montre que les fraudes ne sont pas uniformément petites : beaucoup de montants très faibles, et une partie de montants moyens ou élevés qui tirent la moyenne.
- Les fraudes se concentrent sur quelques montants précis. 22 % des fraudes (105 sur 473) sont à exactement 1,00 €, contre 4,8 % des transactions normales. D’autres montants reviennent aussi (99,99 € pour 27 fraudes, 0,00 € pour 25).
- Les 27 fraudes à 99,99 € se sont produites en environ 7 heures, avec un écart médian de 33 secondes entre deux transactions. Cela est compatible avec une opération répétée, sans le prouver, car le dataset ne contient aucun identifiant de carte. >
